```python
import pandas as pd

aerolines = pd.read_csv('train.csv')
display(aerolines)

# siendo "satisfaction" el objetivo
# me dice cuántos hay de cada tipo en dicha columna
class_distribution = aerolines["satisfaction"].value_counts()
print(class_distribution)

countNaN = aerolines.isna().sum()
print("=== countNaN ===")
print(countNaN)

# Se opta por borrar las filas que tengan uno o más datos con NaN
# Se pueden rellenar con la media o con la mediana
# Pero se ha decidido así pq hay 100.000 filas y solo 300 NaN
aerolines = aerolines.dropna(how='any')
countNaN = aerolines.isna().sum()
print("=== countNaN ===")
print(countNaN)
```

```python
# Borra la columna "Unamed: 0" y "id"
aerolines = aerolines.drop("Unnamed: 0",axis=1)
aerolines = aerolines.drop("id",axis=1)
display(aerolines)
```

ONE HOT ENCODING

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)

Gender=aerolines["Gender"]
GenderOHE = encoder.fit_transform(Gender.to_numpy().reshape(-1, 1))
aerolines = aerolines.drop("Gender",axis=1)

CustomerType=aerolines["Customer Type"]
CustomerTypeOHE = encoder.fit_transform(CustomerType.to_numpy().reshape(-1, 1))
aerolines = aerolines.drop("Customer Type",axis=1)

TypeofTravel=aerolines["Type of Travel"]
TypeofTraveOHE = encoder.fit_transform(TypeofTravel.to_numpy().reshape(-1, 1))
aerolines = aerolines.drop("Type of Travel",axis=1)

Class=aerolines["Class"]
ClassOHE = encoder.fit_transform(Class.to_numpy().reshape(-1, 1))
aerolines = aerolines.drop("Class",axis=1)

display(aerolines)
print(GenderOHE) #tambien valdria labelEncoder.
```

```python
import numpy as np
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

YPandas = aerolines["satisfaction"]     # saca la columna
print(YPandas)
print(YPandas.to_numpy())

Y = le.fit_transform(YPandas.to_numpy())    # lo transforma en 1 y 0
print(Y)

aerolines = aerolines.drop("satisfaction",axis=1)   # quita la Y
X = aerolines.to_numpy()                            # tenemos la X
print(X.shape)

X = np.hstack((aerolines,GenderOHE,CustomerTypeOHE,TypeofTraveOHE,ClassOHE))    # le ponemos las columnas que habiamos hecho ONE HOT ENCODING
print(X.shape)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

scaler = StandardScaler()
X = scaler.fit_transform(X)                             # escala todo 
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.25, random_state=0)       # separa en 25 / 75
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

MLP

```python
# ReLU -> no linear, dos capas de neuronas ocultas(100 y 20), alpha = 0.01 aprende lento, learning_rate_init = velocidad de aprendizaje
model = MLPClassifier(activation='relu',hidden_layer_sizes=(100,20), max_iter=7000, alpha=0.01, learning_rate_init=0.01, random_state=0)
model.fit(X_train, y_train)
# Obtener las predicciones del clasificador de scikit-learn
y_pred_sklearn = model.predict(X_test)

ac = accuracy_score(y_pred_sklearn,y_test)
print("Accuracy",ac)
```

RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier(n_estimators=1000,max_depth=20, random_state=0)
clf.fit(X_train, y_train)
y_predictRF = clf.predict(X_test)

rf_accuracy = accuracy_score(y_test, y_predictRF)
print("accuracy",rf_accuracy)
```

```python

# Por defecto tiene criterian = "gini" , siendo gini = 0 que estás 100% clasificado y 0.5 mitad y mitad
#criterion = "entropy" usa la formula de los logaritmos esta, es más lenta
clf = RandomForestClassifier(criterion="entropy", n_estimators=1000,max_depth=20, random_state=0)
clf.fit(X_train, y_train)
y_predictRF = clf.predict(X_test)

rf_accuracy = accuracy_score(y_test, y_predictRF)
print("accuracy",rf_accuracy)
```
