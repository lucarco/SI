```python
import pandas as pd
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
import numpy as np



cat_breeds = pd.read_csv('cat_breeds_clean.csv',sep=",")
display(cat_breeds)
imputer = SimpleImputer(strategy='median')  # para rellenar los huecos de datos vacios con la mediana

# Simplemente lo ves con tus propios ojos cuales faltan
# convierte todo lo de la columna "Latitude" que sea "-" y lo convierte a NaN (Not a number)
cat_breeds["Latitude"]=cat_breeds["Latitude"].apply(lambda x: np.nan if x == "-" else x)

# Suma todos los NaN
countNaN = cat_breeds.isna().sum()
print("=== Todos los countNaN ===")
print(countNaN)


# Codifica "Gender" 
Gender = cat_breeds["Gender"].to_numpy()    # la extrae de esta manera para que las biblios puedan trabajar con ella
labelEncoder=LabelEncoder()
Gender = labelEncoder.fit_transform(Gender)     # hace male = 0 y female = 1 y unknown = 2 (por ejemplo)
# rellena los huecos vacios con la mediana (imputer declarado al principio)
# reshape: lo pone todo en una sola columna(1) y pone tantas filas como sean necesarias(-1)
Gender=imputer.fit_transform(Gender.reshape(-1, 1))     
cat_breeds["Gender"] = Gender   # sobreescribe en el df original

# creas un imputer pero esta vez con la media
imputerMean = SimpleImputer(strategy='mean')
Latitude=cat_breeds["Latitude"].to_numpy()
Latitude=imputerMean.fit_transform(Latitude.reshape(-1, 1))
cat_breeds["Latitude"] = Latitude

countNaN = cat_breeds.isna().sum()
print("=== LOS COUNTNAN NUEVAMENTE ===")
print(countNaN)
display(cat_breeds.head())
```

```python
#Label Encoding

# Neutered_or_spayed: True / False
Neutered_or_spayed = cat_breeds["Neutered_or_spayed"]                   # extraer columna
Neutered_or_spayed_NP = Neutered_or_spayed.to_numpy()                   # convertir a formato matematico
Neutered_or_spayed = labelEncoder.fit_transform(Neutered_or_spayed_NP)     # transformar palabras a numeros
cat_breeds["Neutered_or_spayed"] = Neutered_or_spayed                   

# Allowed_outdoor: True / False
Allowed_outdoor = cat_breeds["Allowed_outdoor"]
Allowed_outdoor_NP = Allowed_outdoor.to_numpy()
Allowed_outdoor_NP = labelEncoder.fit_transform(Allowed_outdoor_NP)
cat_breeds["Allowed_outdoor"] = Allowed_outdoor_NP

# Preferred_food: dry /wet
Preferred_food = cat_breeds["Preferred_food"]
Preferred_food_NP = Preferred_food.to_numpy()
Preferred_food_NP = labelEncoder.fit_transform(Preferred_food_NP)
cat_breeds["Preferred_food"] = Preferred_food_NP




# ONE HOT Encoding.
# sparse_output = False . Es para que sea una matriz de 1 y 0 en vez de un formato comprimido raro
encoder = OneHotEncoder(sparse_output=False)

Fur_colour_dominant=cat_breeds["Fur_colour_dominant"]                                       # extraer columna
Fur_colour_dominant_NP = Fur_colour_dominant.to_numpy()                                     # convertir en formato matemático
Fur_colour_dominant_OHE = encoder.fit_transform(Fur_colour_dominant_NP.reshape(-1, 1))      # crea la nueva matriz de 1 y 0
cat_breeds = cat_breeds.drop("Fur_colour_dominant",axis=1)                                  # borra la columa original

Fur_pattern=cat_breeds["Fur_pattern"]
Fur_pattern_NP = Fur_pattern.to_numpy()
Fur_pattern_NP_OHE = encoder.fit_transform(Fur_pattern_NP.reshape(-1, 1))
cat_breeds = cat_breeds.drop("Fur_pattern",axis=1)

Eye_colour=cat_breeds["Eye_colour"]
Eye_colour_NP = Eye_colour.to_numpy()
Eye_colour_NP_OHE = encoder.fit_transform(Eye_colour_NP.reshape(-1, 1))
cat_breeds = cat_breeds.drop("Eye_colour",axis=1)

Country=cat_breeds["Country"]
Country_NP = Country.to_numpy()
Country_NP_OHE = encoder.fit_transform(Country_NP.reshape(-1, 1))
cat_breeds = cat_breeds.drop("Country",axis=1)

display(cat_breeds.head())
```

```python
# Coges la Y (el objetivo) y lo borras de la tabla principal
Y = cat_breeds["Breed"].to_numpy()
cat_breeds=cat_breeds.drop("Breed",axis=1)

# Coges la X como las pistas o las características
X = cat_breeds.to_numpy()
print(X.shape)

# Le pones a la X todo lo que has hecho con HOT ENCODING
X = np.hstack((X,Fur_colour_dominant_OHE,Fur_pattern_NP_OHE,Eye_colour_NP_OHE,Country_NP_OHE))
print(X.shape)
```

MVP

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Escalar para que todo la media 0 y desviacion tipica 1
scaler = StandardScaler()
X = scaler.fit_transform(X)                 # escala las características
Y = labelEncoder.fit_transform(Y)           # escala el objetivo
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.25, random_state=0)       # divide en 25% para el examen final

print("y_train")
print(y_train)
print("X_train.shape")
print(X_train.shape)
print("X_test.shape")
print(X_test.shape)
print("y_train.shape")
print(y_train.shape)
print("y_test.shape")
print(y_test.shape)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Tamaño de X_train:", X_train.shape)
print("Tamaño de y_train:", y_train.shape)

# 1. Crear el modelo (Usamos los mismos parámetros que tu profesor puso antes)
mlp = MLPClassifier(
    hidden_layer_sizes=(30, 10), # Dos capas ocultas con 30 y 10 neuronas
    activation='relu',           # Función no lineal
    max_iter=7000,               # Límite de épocas de estudio
    alpha=0.01,                  # Regularización (evita el sobreajuste)
    learning_rate_init=0.001,    # Velocidad de aprendizaje
    random_state=0               # Semilla para que el resultado sea siempre el mismo
)

# 2. Entrenar la Red Neuronal
print("\nEntrenando la red neuronal... (esto puede tardar unos segundos)")
mlp.fit(X_train, y_train)

# 3. Hacer las predicciones con el conjunto de examen (test)
y_pred = mlp.predict(X_test)

# 4. Evaluar los resultados
accuracy = accuracy_score(y_test, y_pred)

print("\n=== RESULTADOS DEL MODELO ===")
print("Accauricy: ", accuracy)
print(f"Accuracy (Porcentaje de acierto): {accuracy * 100:.2f}%\n")

# Extra: Añadimos la matriz de confusión y el reporte detallado que siempre viene bien
print("Matriz de Confusión:")
print(confusion_matrix(y_test, y_pred))
```

DECISION TREE

```python
from sklearn.tree import DecisionTreeClassifier

depth = []
accDT = []

for i in range(1,20):
    decision_tree = DecisionTreeClassifier(random_state=0, max_depth=i)
    # Entrenar modelo
    decision_tree.fit(X_train, y_train)
    # Hacer prediccion
    dt_pred = decision_tree.predict(X_test)
    # Accuracy
    dt_accuracy = accuracy_score(y_test, dt_pred)
    print("max_depth ",i,"Acc ",dt_accuracy)
    depth.append(i)
    accDT.append(dt_accuracy)
    
dt_accuracyMax = np.argmax(accDT)
print("Accuracy ",accDT[dt_accuracyMax], " with max_depth ",depth[dt_accuracyMax])
```

RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier

# n_estimators = número de árboles de decisiones
clf = RandomForestClassifier(n_estimators=2000,max_depth=9, random_state=0) # construir
clf.fit(X_train, y_train)               # entrenar
y_predictRF = clf.predict(X_test)       # predecir

rf_accuracy = accuracy_score(y_test, y_predictRF)
print("accuracy",rf_accuracy)
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier

ks = []                     # las k
acc = []                    # la accuaricy de cada k
knnPredictions = []         # las predicciones de cada k

# Probar k E [1, 20]
for i in range(1,20):
    
    # n_neighbors = la k del momento , metric = la metrica que quieras usar
    neigh = KNeighborsClassifier(n_neighbors=i, metric="manhattan")
    neigh.fit(X_train, y_train)                                 # entreno
    y_predictKNN = neigh.predict(X_test)                        # prediccion
    knn_accuracy = accuracy_score(y_test, y_predictKNN)         # accuaracy

    knnPredictions.append(y_predictKNN)
    ks.append(i)
    acc.append(knn_accuracy)
    print("K ",i,knn_accuracy)

knnAccuracyMax = np.argmax(acc)
print("Accuracy ",acc[knnAccuracyMax], " with K ",ks[knnAccuracyMax])
```
